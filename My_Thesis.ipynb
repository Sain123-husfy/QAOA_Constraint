```python
%matplotlib inline
import numpy as np
import random
import networkx as nx
import matplotlib.pyplot as plt
from qiskit.circuit.library import QAOAAnsatz
from qiskit.quantum_info import SparsePauliOp
from qiskit_aer.primitives import Sampler
from qiskit.visualization import plot_histogram
from itertools import product
from scipy.optimize import minimize

# ============================================================
#  Set random seeds for reproducibility
# ============================================================
np.random.seed(42)
random.seed(42)

# ============================================================
#  USER SETTINGS – your graph
# ============================================================
nodes = [0,1,2,3,4]
edges = [(0,1), (1,2), (2,3), (3,4), (4,0), (0,3)]
n_qubits = len(nodes)

# QAOA settings
reps_list = [1,2,3,4,5]          # layers to test
grid_resolution = 32             # grid points for p=1
shots = 5000                     # increased for better accuracy

# ============================================================
#  Build MaxCut Hamiltonian
# ============================================================
def maxcut_hamiltonian(edges, n):
    terms = []
    for i, j in edges:
        z_string = ['I'] * n
        z_string[i] = 'Z'
        z_string[j] = 'Z'
        pauli_zz = "".join(z_string)
        terms.append((pauli_zz, 0.5))   # +0.5 * Z_i Z_j
        terms.append(("I"*n, -0.5))     # -0.5 * I
    return SparsePauliOp.from_list(terms)

hamiltonian = maxcut_hamiltonian(edges, n_qubits)
print("Hamiltonian:")
print(hamiltonian)

# ============================================================
#  Helper: compute cut & energy from counts
# ============================================================
def compute_energy_and_cut(counts, edges):
    avg_cut = 0.0
    for bits, prob in counts.items():
        cut = 0
        for u, v in edges:
            if bits[u] != bits[v]:
                cut += 1
        avg_cut += prob * cut
    return -avg_cut, avg_cut   # energy = -cut

# ============================================================
#  Initialize sampler
# ============================================================
sampler = Sampler()

# ============================================================
#  Storage for results
# ============================================================
all_results = {}

# ============================================================
#  QAOA loop over p = reps
# ============================================================
for reps in reps_list:
    print("\n" + "="*60)
    print(f"Processing reps = {reps} (parameter dimension = {2*reps})")
    print("="*60)

    ansatz = QAOAAnsatz(hamiltonian, reps=reps)

    best_energy = float('inf')
    best_params = None
    best_counts = None
    best_cut = None

    if reps == 1:
        # --------------------------------------------------------
        # Grid search for p=1 (exhaustive)
        # --------------------------------------------------------
        print("Performing grid search...")
        gamma_vals = np.linspace(0, 2*np.pi, grid_resolution)
        beta_vals = np.linspace(0, 2*np.pi, grid_resolution)
        total_points = grid_resolution * grid_resolution
        evaluated = 0

        for gamma, beta in product(gamma_vals, beta_vals):
            params = [gamma, beta]
            circuit = ansatz.assign_parameters(params)
            circuit.measure_all()

            job = sampler.run([circuit], shots=shots)
            result = job.result()
            counts = result.quasi_dists[0].binary_probabilities()
            energy, cut = compute_energy_and_cut(counts, edges)

            if energy < best_energy:
                best_energy = energy
                best_params = params
                best_counts = counts
                best_cut = cut

            evaluated += 1
            if evaluated % 50 == 0:
                print(f"  Evaluated {evaluated}/{total_points} points")

    else:
        # --------------------------------------------------------
        # Use COBYLA optimizer with multiple restarts for p>1
        # --------------------------------------------------------
        print(f"Optimizing with COBYLA (multiple restarts)...")
        
        def objective(params):
            circuit = ansatz.assign_parameters(params)
            circuit.measure_all()
            job = sampler.run([circuit], shots=shots)
            result = job.result()
            counts = result.quasi_dists[0].binary_probabilities()
            energy, _ = compute_energy_and_cut(counts, edges)
            return energy

        num_restarts = 15   # try 15 different starting points
        for restart in range(num_restarts):
            init_params = np.random.uniform(0, 2*np.pi, 2*reps).tolist()
            res = minimize(objective, init_params, method='COBYLA',
                           options={'maxiter': 500})
            # Evaluate final circuit to get cut
            final_circuit = ansatz.assign_parameters(res.x)
            final_circuit.measure_all()
            job = sampler.run([final_circuit], shots=shots)
            final_counts = job.result().quasi_dists[0].binary_probabilities()
            final_energy, final_cut = compute_energy_and_cut(final_counts, edges)
            
            if final_energy < best_energy:
                best_energy = final_energy
                best_params = res.x
                best_counts = final_counts
                best_cut = final_cut
            if (restart+1) % 5 == 0:
                print(f"  Restart {restart+1}/{num_restarts}: best cut so far = {best_cut:.4f}")

    # Store results
    all_results[reps] = {
        'best_params': best_params,
        'best_energy': best_energy,
        'best_cut': best_cut,
        'best_counts': best_counts
    }

    print("\n>>> Best parameters found:")
    if reps == 1:
        print(f"    γ = {best_params[0]:.4f}, β = {best_params[1]:.4f}")
    else:
        param_str = ", ".join([f"{p:.4f}" for p in best_params])
        print(f"    parameters = [{param_str}]")
    print(f"    Minimum energy = {best_energy:.6f}")
    print(f"    Maximum cut value = {best_cut:.4f} out of {len(edges)} edges")

    # Plot histogram with overlapping fix
    plt.figure(figsize=(12, 6))
    plot_histogram(best_counts, title=f"Best QAOA distribution (reps={reps})")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print("Probabilities (bitstring → cut value):")
    for bits, prob in sorted(best_counts.items(), key=lambda x: -x[1])[:10]:  # show top 10
        cut_val = 0
        for u, v in edges:
            if bits[u] != bits[v]:
                cut_val += 1
        print(f"  |{bits}⟩: {prob:.4f}  (cut = {cut_val})")

# ============================================================
#  Summary of all results
# ============================================================
print("\n" + "="*60)
print("SUMMARY OF ALL RESULTS")
print("="*60)
for reps, data in all_results.items():
    print(f"reps = {reps}: best cut = {data['best_cut']:.4f} / {len(edges)} edges")

# ============================================================
#  Save the p=5 histogram as compressed TIFF
# ============================================================
if 5 in all_results:
    counts_p5 = all_results[5]['best_counts']
    fig, ax = plt.subplots(figsize=(12, 6))
    plot_histogram(counts_p5, ax=ax, title="Best QAOA distribution (reps=5)")
    ax.tick_params(axis='x', rotation=45)
    plt.tight_layout()
    plt.savefig('qaoa_p5_histogram.tif', format='tiff', dpi=300, bbox_inches='tight',
                pil_kwargs={'compression': 'tiff_lzw'})
    plt.show()
    print("Saved histogram as 'qaoa_p5_histogram.tif' (compressed LZW)")
else:
    print("No results for p=5 found.")

# ============================================================
#  Compute exact Max-Cut and plot approximation ratio vs p
# ============================================================
def exact_max_cut_bruteforce(edges, n):
    max_cut = 0
    for mask in range(2**n):
        cut = 0
        for u, v in edges:
            if ((mask >> u) & 1) != ((mask >> v) & 1):
                cut += 1
        max_cut = max(max_cut, cut)
    return max_cut

exact_cut = exact_max_cut_bruteforce(edges, n_qubits)
print(f"\nExact Max-Cut = {exact_cut} out of {len(edges)} edges")

p_vals = sorted(all_results.keys())
ratios = [all_results[p]['best_cut'] / exact_cut for p in p_vals]

plt.figure(figsize=(8,5))
plt.plot(p_vals, ratios, 'bo-', linewidth=2, markersize=8)
plt.xlabel('Number of QAOA layers (p)', fontsize=12)
plt.ylabel('Approximation ratio', fontsize=12)
plt.title('QAOA approximation ratio vs. p', fontsize=14)
plt.grid(True, alpha=0.3)
plt.ylim(0, 1.05)
for i, (p, r) in enumerate(zip(p_vals, ratios)):
    plt.annotate(f'{r:.4f}', (p, r), textcoords="offset points", xytext=(0,10), ha='center')
plt.savefig('qaoa_ratio_vs_p.tif', format='tiff', dpi=300, bbox_inches='tight',
            pil_kwargs={'compression': 'tiff_lzw'})
plt.show()
print("Saved ratio plot as 'qaoa_ratio_vs_p.tif' (compressed LZW)")
```

```python
# ============================================================
#  Complete Custom RQAOA for MaxCut (Efficient & Self‑Contained)
# ============================================================
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.quantum_info import Statevector, SparsePauliOp
from scipy.optimize import minimize

# -------------------------------
# 1. Graph definition (modify here)
# -------------------------------
nodes = [0, 1, 2, 3, 4]
edges = [(0, 1), (1, 2), (2, 3), (3, 4), (4, 0), (0, 3)]
n_qubits = len(nodes)

# -------------------------------
# 2. QAOA circuit (works for any number of layers)
# -------------------------------
def qaoa_circuit(beta_list, gamma_list, edges, n):
    qc = QuantumCircuit(n)
    qc.h(range(n))
    reps = len(beta_list)
    for layer in range(reps):
        # Cost layer (ZZ rotations)
        for u, v in edges:
            qc.cx(u, v)
            qc.rz(2 * gamma_list[layer], v)
            qc.cx(u, v)
        # Mixer layer (X rotations)
        for i in range(n):
            qc.rx(2 * beta_list[layer], i)
    return qc

# -------------------------------
# 3. Build MaxCut Hamiltonian (for any subgraph)
# -------------------------------
def build_hamiltonian(edges, n):
    terms = []
    for i, j in edges:
        z_str = ['I'] * n
        z_str[i] = 'Z'
        z_str[j] = 'Z'
        pauli = ''.join(z_str)
        terms.append((pauli, 0.5))   # +0.5 * Z_i Z_j
        terms.append(('I'*n, -0.5))  # -0.5 * I
    return SparsePauliOp.from_list(terms)

# -------------------------------
# 4. Optimize QAOA (multiple restarts, statevector simulation)
# -------------------------------
def optimize_qaoa(hamiltonian, edges, n_qubits, reps=2, num_restarts=15, maxiter=500):
    backend = Aer.get_backend('aer_simulator_statevector')
    
    def objective(params):
        gamma_list = params[:reps]
        beta_list = params[reps:]
        qc = qaoa_circuit(beta_list, gamma_list, edges, n_qubits)
        state = Statevector.from_instruction(qc)
        return np.real(state.expectation_value(hamiltonian))
    
    best_energy = float('inf')
    best_params = None
    best_state = None
    for restart in range(num_restarts):
        init = np.random.uniform(0, 2*np.pi, 2*reps).tolist()
        res = minimize(objective, init, method='COBYLA', options={'maxiter': maxiter})
        if res.fun < best_energy:
            best_energy = res.fun
            best_params = res.x
            gamma_list = best_params[:reps]
            beta_list = best_params[reps:]
            best_state = Statevector.from_instruction(qaoa_circuit(beta_list, gamma_list, edges, n_qubits))
        if (restart+1) % 5 == 0:
            print(f"    Restart {restart+1}/{num_restarts}: best energy = {best_energy:.6f}")
    return best_params, best_energy, best_state

# -------------------------------
# 5. Compute correlations <Z_i> and <Z_i Z_j>
# -------------------------------
def compute_correlations(state, n, edges):
    corr_z = {}
    corr_zz = {}
    for i in range(n):
        z_str = ['I'] * n
        z_str[i] = 'Z'
        op = SparsePauliOp.from_list([(''.join(z_str), 1.0)])
        corr_z[i] = np.real(state.expectation_value(op))
    for i, j in edges:
        z_str = ['I'] * n
        z_str[i] = 'Z'
        z_str[j] = 'Z'
        op = SparsePauliOp.from_list([(''.join(z_str), 1.0)])
        corr_zz[(i, j)] = np.real(state.expectation_value(op))
    return corr_z, corr_zz

# -------------------------------
# 6. Elimination based on largest |<Z_i Z_j>|
# -------------------------------
def eliminate_zz(corr_zz, n, edges):
    best_edge = max(corr_zz.items(), key=lambda x: abs(x[1]))[0]
    i, j = best_edge
    relation = 'same' if corr_zz[best_edge] > 0 else 'opposite'
    return i, j, relation

# -------------------------------
# 7. Reduce graph after eliminating a node
# -------------------------------
def reduce_graph(edges, n, eliminated_node, kept_node=None, relation=None):
    if kept_node is not None:
        # eliminate 'eliminated_node' by setting it equal/opposite to 'kept_node'
        new_edges = []
        for u, v in edges:
            u2 = kept_node if u == eliminated_node else u
            v2 = kept_node if v == eliminated_node else v
            if u2 == v2:
                continue
            new_edges.append((min(u2, v2), max(u2, v2)))
        new_edges = list(set(new_edges))
        remaining = [node for node in range(n) if node != eliminated_node]
        return new_edges, remaining
    else:
        # simply remove the node (used for fixed-value elimination, not used here)
        new_edges = [(u, v) for u, v in edges if u != eliminated_node and v != eliminated_node]
        remaining = [node for node in range(n) if node != eliminated_node]
        return new_edges, remaining

# -------------------------------
# 8. Local search to improve final cut
# -------------------------------
def local_search(assignment, edges):
    n = len(assignment)
    current_cut = sum(1 for u, v in edges if assignment[u] != assignment[v])
    improved = True
    while improved:
        improved = False
        for i in range(n):
            assignment[i] = 1 - assignment[i]
            new_cut = sum(1 for u, v in edges if assignment[u] != assignment[v])
            if new_cut > current_cut:
                current_cut = new_cut
                improved = True
            else:
                assignment[i] = 1 - assignment[i]
    return assignment, current_cut

# -------------------------------
# 9. Main RQAOA routine (uses ZZ elimination)
# -------------------------------
def rqaoa(original_edges, n_qubits, reps=2, num_restarts=15, min_size=2):
    current_edges = original_edges[:]
    current_n = n_qubits
    node_map = list(range(n_qubits))          # maps current index -> original node
    elim_history = []                         # (elim_node, keep_node, relation)
    
    while current_n > min_size:
        print(f"\n--- RQAOA iteration: {current_n} nodes left ---")
        H = build_hamiltonian(current_edges, current_n)
        _, _, state = optimize_qaoa(H, current_edges, current_n,
                                    reps=reps, num_restarts=num_restarts)
        _, corr_zz = compute_correlations(state, current_n, current_edges)
        i, j, relation = eliminate_zz(corr_zz, current_n, current_edges)
        # Keep i, eliminate j
        new_edges, remaining = reduce_graph(current_edges, current_n,
                                            eliminated_node=j, kept_node=i, relation=relation)
        orig_elim = node_map[j]
        orig_keep = node_map[i]
        elim_history.append((orig_elim, orig_keep, relation))
        print(f"  Eliminated node {orig_elim} based on {relation} correlation with {orig_keep} (|corr|={abs(corr_zz[(i,j)]):.4f})")
        
        # Renumber remaining nodes
        old_to_new = {old: new for new, old in enumerate(remaining)}
        new_edges_renum = [(old_to_new[u], old_to_new[v]) for u, v in new_edges]
        current_edges = new_edges_renum
        current_n = len(remaining)
        node_map = [node_map[old] for old in remaining]
    
    # Solve final small graph exactly (brute force)
    print(f"\n--- Solving final graph of size {current_n} by brute force ---")
    best_cut_small = -1
    best_assign_small = None
    for mask in range(2**current_n):
        assign = {i: (mask >> i) & 1 for i in range(current_n)}
        cut = sum(1 for u, v in current_edges if assign[u] != assign[v])
        if cut > best_cut_small:
            best_cut_small = cut
            best_assign_small = assign
    # Map back to original node indices
    final_assign = {}
    for new_idx, val in best_assign_small.items():
        orig_node = node_map[new_idx]
        final_assign[orig_node] = val
    # Backtrack elimination history
    for elim_node, keep_node, relation in reversed(elim_history):
        if relation == 'same':
            final_assign[elim_node] = final_assign[keep_node]
        else:
            final_assign[elim_node] = 1 - final_assign[keep_node]
    # Build full assignment list
    full_assignment = [final_assign[i] for i in range(n_qubits)]
    # Local search refinement
    full_assignment, final_cut = local_search(full_assignment, original_edges)
    return final_cut, full_assignment

# -------------------------------
# 10. Exact Max‑Cut (brute force) for comparison
# -------------------------------
def exact_max_cut(edges, n):
    best = 0
    for mask in range(2**n):
        cut = sum(1 for u, v in edges if ((mask >> u) & 1) != ((mask >> v) & 1))
        best = max(best, cut)
    return best

exact_cut = exact_max_cut(edges, n_qubits)
print(f"Exact Max-Cut = {exact_cut} out of {len(edges)} edges")

# -------------------------------
# 11. Run RQAOA
# -------------------------------
print("\n" + "="*60)
print("Running Custom RQAOA (reps=2, 15 restarts, elimination='zz')")
print("="*60)

rqaoa_cut, rqaoa_solution = rqaoa(edges, n_qubits, reps=2, num_restarts=15, min_size=2)

print("\n" + "="*60)
print("FINAL RQAOA RESULT")
print("="*60)
print(f"Solution bitstring: {''.join(str(b) for b in rqaoa_solution)}")
print(f"Cut value: {rqaoa_cut} out of {len(edges)} edges")
print(f"Approximation ratio: {rqaoa_cut / exact_cut:.6f}")
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from scipy.optimize import minimize

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram

# =====================================================================
# 1. SETUP GRAPH & PROBLEM ENVIRONMENT
# =====================================================================
# Create a 6-cycle (C6) graph
num_nodes = 6
G = nx.cycle_graph(num_nodes)
edges = list(G.edges())

# Analytical maximum cut for C6 is exactly 6 (since it is bipartite)
max_cut_val = 6.0

def compute_string_cut(bitstring, graph):
    """Computes the cut value of a specific classical bitstring."""
    # Reverse string to match Qiskit's little-endian bit ordering
    bits = [int(char) for char in bitstring[::-1]]
    cut = 0
    for u, v in graph.edges():
        if bits[u] != bits[v]:
            cut += 1
    return cut

# =====================================================================
# 2. DEFINE PARAMETERIZED QAOA CIRCUIT
# =====================================================================
def create_qaoa_circuit(graph, p, gamma, beta):
    """Builds a gate-wise QAOA circuit for a given depth p and angles."""
    num_nodes = len(graph.nodes())
    qc = QuantumCircuit(num_nodes)
    
    # Step A: Initial state - Apply Hadamard gates to create uniform superposition
    qc.h(range(num_nodes))
    
    # Step B: Alternating Layers of Cost and Mixer operators
    for layer in range(p):
        qc.barrier()
        # --- Cost Hamiltonian Unitary ---
        # U(C, gamma) = e^(-i * gamma * H_C) where H_C = 0.5 * sum(I - ZiZj)
        for u, v in graph.edges():
            qc.cx(u, v)
            qc.rz(2 * gamma[layer], v)
            qc.cx(u, v)
            
        qc.barrier()
        # --- Mixer Hamiltonian Unitary ---
        # U(B, beta) = e^(-i * beta * H_B) where H_B = sum(X_i)
        for node in graph.nodes():
            qc.rx(2 * beta[layer], node)
            
    return qc

# =====================================================================
# 3. OPTIMIZATION & EXPECTATION VALUE ENERGY EVALUATION
# =====================================================================
def evaluate_expectation(params, graph, p):
    """Calculates the negative average cut value for classical minimization."""
    # Split the parameter array into gamma and beta sets
    gamma = params[:p]
    beta = params[p:]
    
    qc = create_qaoa_circuit(graph, p, gamma, beta)
    state = Statevector.from_instruction(qc)
    probabilities = state.probabilities_dict()
    
    avg_cut = 0
    for bitstring, prob in probabilities.items():
        cut = compute_string_cut(bitstring, graph)
        avg_cut += cut * prob
        
    # Return negative value because scipy minimizes by default
    return -avg_cut

# =====================================================================
# 4. EXECUTION LOOP ACROSS DEPTHS p = 1 TO 4
# =====================================================================
approximation_ratios = []
optimal_params_history = {}

print("--- Running Variational Optimization Loops ---")
for p in range(1, 5):
    # Initialize random starting points for angles
    initial_guess = np.random.uniform(-np.pi, np.pi, 2 * p)
    
    # Optimize parameters using COBYLA
    res = minimize(
        evaluate_expectation, 
        initial_guess, 
        args=(G, p), 
        method='COBYLA', 
        options={'maxiter': 500}
    )
    
    # Extract the maximized objective value
    optimized_avg_cut = -res.fun
    ratio = optimized_avg_cut / max_cut_val
    approximation_ratios.append(ratio)
    optimal_params_history[p] = res.x
    
    print(f"Depth p = {p} completed. Approximation Ratio r = {ratio:.4f}")

# =====================================================================
# 5. VISUALIZATION & ANALYSIS PLOTS
# =====================================================================
print("\nGenerating layout visualizations...")

# --- Plot 1: Trend of Approximation Ratio ---
plt.figure(figsize=(6, 4))
plt.plot(range(1, 5), approximation_ratios, marker='o', color='darkviolet', linewidth=2)
plt.title("QAOA Approximation Ratio Scaling on $C_6$")
plt.xlabel("Circuit Depth ($p$)")
plt.ylabel("Approximation Ratio ($r$)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(range(1, 5))
plt.ylim(0.5, 1.05)
plt.show()

# --- Plot 2: Probability Distribution at Final Depth (p=4) ---
best_p4_params = optimal_params_history[4]
p4_gamma = best_p4_params[:4]
p4_beta = best_p4_params[4:]
final_qc = create_qaoa_circuit(G, 4, p4_gamma, p4_beta)

final_state = Statevector.from_instruction(final_qc)
p4_probabilities = final_state.probabilities_dict()

# Retain only bitstrings with meaningful output weight for visual clarity
filtered_probs = {k: v for k, v in p4_probabilities.items() if v > 0.01}
plot_histogram(filtered_probs, title="Final Bitstring Distribution (Depth p=4)")
plt.show()

# --- Plot 3: Gate-Wise Circuit Structure Structure ---
print("\n--- Gate-wise Circuit Matrix (Depth p=4) ---")
# Change the renderer to 'mpl' (Matplotlib) and set a folding limit
# 'fold=25' tells Qiskit how many gates to place horizontally before wrapping to a new line row
final_qc.draw(output='mpl', fold=25)

# Display the plot window
plt.show()
```

```python
import time
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, SparsePauliOp

# ------------------------------------------------------------
# 1. Graph setup
# ------------------------------------------------------------
G_full = nx.cycle_graph(4)
optimal_cut = G_full.number_of_edges()   # = 4

# ------------------------------------------------------------
# 2. QAOA circuit and expectation (works for any graph)
# ------------------------------------------------------------
def qaoa_circuit(graph, gamma, beta):
    nodes = list(graph.nodes())
    node_to_qubit = {node: i for i, node in enumerate(nodes)}
    qc = QuantumCircuit(len(nodes))
    qc.h(range(len(nodes)))
    for u, v in graph.edges():
        qu = node_to_qubit[u]
        qv = node_to_qubit[v]
        qc.cx(qu, qv)
        qc.rz(2 * gamma, qv)
        qc.cx(qu, qv)
    qc.rx(2 * beta, range(len(nodes)))
    return qc

def maxcut_expectation(statevector, graph):
    nodes = list(graph.nodes())
    node_to_qubit = {node: i for i, node in enumerate(nodes)}
    exp = 0.0
    for u, v in graph.edges():
        qu = node_to_qubit[u]
        qv = node_to_qubit[v]
        pauli = ['I'] * len(nodes)
        pauli[qu] = 'Z'
        pauli[qv] = 'Z'
        op = SparsePauliOp.from_list([(''.join(pauli[::-1]), 1.0)])
        zz = np.real(statevector.expectation_value(op))
        exp += (1 - zz) / 2.0
    return exp

def correlations(statevector, graph):
    nodes = list(graph.nodes())
    node_to_qubit = {node: i for i, node in enumerate(nodes)}
    corr = {}
    for u, v in graph.edges():
        qu = node_to_qubit[u]
        qv = node_to_qubit[v]
        pauli = ['I'] * len(nodes)
        pauli[qu] = 'Z'
        pauli[qv] = 'Z'
        op = SparsePauliOp.from_list([(''.join(pauli[::-1]), 1.0)])
        corr[(u, v)] = np.real(statevector.expectation_value(op))
    return corr

# ------------------------------------------------------------
# 3. Optimise QAOA angles (p=1)
# ------------------------------------------------------------
def objective(params, graph):
    gamma, beta = params
    qc = qaoa_circuit(graph, gamma, beta)
    sv = Statevector.from_instruction(qc)
    return -maxcut_expectation(sv, graph)

print("Optimising QAOA angles...")
res = minimize(objective, x0=[0.5, 0.5], args=(G_full,), method='COBYLA')
gamma_opt, beta_opt = res.x
print(f"Optimal angles: γ = {gamma_opt:.4f}, β = {beta_opt:.4f}")

# ------------------------------------------------------------
# 4. QAOA with timing
# ------------------------------------------------------------
start_qaoa = time.perf_counter()
qc_qaoa = qaoa_circuit(G_full, gamma_opt, beta_opt)
sv_qaoa = Statevector.from_instruction(qc_qaoa)
qaoa_cut = maxcut_expectation(sv_qaoa, G_full)
time_qaoa = time.perf_counter() - start_qaoa
qaoa_ratio = qaoa_cut / optimal_cut

# ------------------------------------------------------------
# 5. RQAOA with timing
# ------------------------------------------------------------
def rqaoa(initial_graph, gamma, beta):
    G_curr = initial_graph.copy()
    parent = {node: node for node in initial_graph.nodes()}
    parity = {node: 0 for node in initial_graph.nodes()}

    def find(node):
        if parent[node] != node:
            root, p = find(parent[node])
            parity[node] ^= p
            parent[node] = root
        return parent[node], parity[node]

    def union(u, v, same_spin):
        ru, pu = find(u)
        rv, pv = find(v)
        if ru == rv:
            return
        new_parity = pu ^ pv ^ (0 if same_spin else 1)
        parent[rv] = ru
        parity[rv] = new_parity

    while G_curr.number_of_nodes() > 1:
        qc = qaoa_circuit(G_curr, gamma, beta)
        sv = Statevector.from_instruction(qc)
        corr = correlations(sv, G_curr)
        (u, v), val = max(corr.items(), key=lambda x: abs(x[1]))
        union(u, v, val > 0)
        G_curr = nx.contracted_nodes(G_curr, u, v, self_loops=False)
        G_curr = nx.relabel.convert_node_labels_to_integers(G_curr)

    assignments = {}
    any_node = next(iter(initial_graph.nodes()))
    root, _ = find(any_node)
    for node in initial_graph.nodes():
        _, p = find(node)
        assignments[node] = p
    return assignments

start_rqaoa = time.perf_counter()
rq_assign = rqaoa(G_full, gamma_opt, beta_opt)
rq_cut = sum(1 for u, v in G_full.edges() if rq_assign[u] != rq_assign[v])
time_rqaoa = time.perf_counter() - start_rqaoa
rq_ratio = rq_cut / optimal_cut

# ------------------------------------------------------------
# 6. Print table of results
# ------------------------------------------------------------
print("\n" + "="*70)
print("COMPARISON TABLE")
print("="*70)
print(f"{'Algorithm':<12} {'Cut value':<12} {'Optimal cut':<12} {'Approx. ratio':<15} {'Time (s)':<10}")
print("-"*70)
print(f"{'QAOA':<12} {qaoa_cut:<12.4f} {optimal_cut:<12} {qaoa_ratio:<15.4f} {time_qaoa:<10.6f}")
print(f"{'RQAOA':<12} {rq_cut:<12} {optimal_cut:<12} {rq_ratio:<15.4f} {time_rqaoa:<10.6f}")
print("="*70)

# ------------------------------------------------------------
# 7. Plots (as requested)
# ------------------------------------------------------------
# QAOA landscape
gamma_vals = np.linspace(0, np.pi, 30)
beta_vals = np.linspace(0, np.pi, 30)
expect = np.zeros((len(gamma_vals), len(beta_vals)))
for i, g in enumerate(gamma_vals):
    for j, b in enumerate(beta_vals):
        qc = qaoa_circuit(G_full, g, b)
        sv = Statevector.from_instruction(qc)
        expect[i, j] = maxcut_expectation(sv, G_full)

plt.figure(figsize=(8, 6))
plt.imshow(expect, extent=[beta_vals[0], beta_vals[-1], gamma_vals[0], gamma_vals[-1]],
           origin='lower', aspect='auto', cmap='viridis')
plt.colorbar(label='⟨C⟩')
plt.scatter(beta_opt, gamma_opt, color='red', marker='*', s=200, label='optimum')
plt.xlabel('β')
plt.ylabel('γ')
plt.title('QAOA (p=1) Expectation Landscape for C4')
plt.legend()
plt.show()

# RQAOA partition
colors = ['red' if rq_assign[node] == 0 else 'blue' for node in G_full.nodes()]
plt.figure(figsize=(5, 4))
pos = nx.circular_layout(G_full)
nx.draw(G_full, pos, node_color=colors, with_labels=True, node_size=500, font_weight='bold')
edge_labels = {(u, v): 'cut' if rq_assign[u] != rq_assign[v] else 'same' for u, v in G_full.edges()}
nx.draw_networkx_edge_labels(G_full, pos, edge_labels=edge_labels, font_size=8)
plt.title(f'RQAOA Partition (Cut size = {rq_cut})')
plt.show()
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

from scipy.optimize import minimize

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# ==========================================================
# PARAMETERS
# ==========================================================

LAMBDA = 2
P = 2                    # QAOA depth
N_RESTARTS = 50          # Random initializations

# ==========================================================
# TEST GRAPHS
# ==========================================================

graphs = {

    "Path $P_3$":
        nx.path_graph(3),

    "Cycle $C_4$":
        nx.cycle_graph(4),

    "Triangle $K_3$":
        nx.complete_graph(3)
}

# ==========================================================
# MIS COST FUNCTION
# ==========================================================

def mis_cost(bitstring, G):

    bits = np.array(
        list(map(int, bitstring))
    )

    reward = np.sum(bits)

    penalty = 0

    for u, v in G.edges():

        penalty += bits[u] * bits[v]

    return reward - LAMBDA * penalty


# ==========================================================
# QAOA CIRCUIT
# ==========================================================

def qaoa_circuit(G, gammas, betas):

    n = len(G.nodes())

    qc = QuantumCircuit(n)

    qc.h(range(n))

    p = len(gammas)

    for layer in range(p):

        gamma = gammas[layer]

        # Reward term

        for q in range(n):
            qc.rz(-gamma, q)

        # Penalty term

        for u, v in G.edges():

            qc.cx(u, v)
            qc.rz(LAMBDA * gamma, v)
            qc.cx(u, v)

        beta = betas[layer]

        # Mixer

        for q in range(n):
            qc.rx(2 * beta, q)

    return qc


# ==========================================================
# EXPECTATION VALUE
# ==========================================================

def expectation(params, G, p):

    gammas = params[:p]
    betas = params[p:]

    qc = qaoa_circuit(
        G,
        gammas,
        betas
    )

    sv = Statevector.from_instruction(qc)

    probs = sv.probabilities()

    value = 0

    n = len(G.nodes())

    for i, prob in enumerate(probs):

        bitstring = format(
            i,
            f"0{n}b"
        )[::-1]

        value += (
            prob *
            mis_cost(bitstring, G)
        )

    return -value


# ==========================================================
# OPTIMIZATION
# ==========================================================

def optimise_qaoa(G, p):

    best_value = 1e9
    best_params = None

    for _ in range(N_RESTARTS):

        x0 = np.random.uniform(
            0,
            np.pi,
            2 * p
        )

        res = minimize(
            expectation,
            x0,
            args=(G, p),
            method="COBYLA",
            options={
                "maxiter": 1000,
                "rhobeg": 0.5
            }
        )

        if res.fun < best_value:

            best_value = res.fun
            best_params = res.x

    return best_params


# ==========================================================
# MAIN ANALYSIS
# ==========================================================

for graph_name, G in graphs.items():

    print("\n" + "=" * 70)
    print(graph_name)
    print("=" * 70)

    n = len(G.nodes())

    # ------------------------------------------------------
    # Optimize Parameters
    # ------------------------------------------------------

    params = optimise_qaoa(G, P)

    gammas = params[:P]
    betas = params[P:]

    print("\nOptimized Parameters")

    for i, g in enumerate(gammas):
        print(f"gamma_{i+1} = {g:.6f}")

    for i, b in enumerate(betas):
        print(f"beta_{i+1}  = {b:.6f}")

    # ------------------------------------------------------
    # Statevector
    # ------------------------------------------------------

    qc = qaoa_circuit(
        G,
        gammas,
        betas
    )

    sv = Statevector.from_instruction(qc)

    probs = sv.probabilities()

    states = [

        format(
            i,
            f"0{n}b"
        )[::-1]

        for i in range(2 ** n)
    ]

    # ------------------------------------------------------
    # Expected MIS Value
    # ------------------------------------------------------

    expected_value = 0

    for state, prob in zip(states, probs):

        expected_value += (
            prob *
            mis_cost(state, G)
        )

    # ------------------------------------------------------
    # Optimal MIS Value
    # ------------------------------------------------------

    optimal_value = max(
        mis_cost(state, G)
        for state in states
    )

    approx_ratio = (
        expected_value /
        optimal_value
    )

    print(
        f"\nExpected MIS Value  = "
        f"{expected_value:.6f}"
    )

    print(
        f"Optimal MIS Value   = "
        f"{optimal_value}"
    )

    print(
        f"Approximation Ratio = "
        f"{approx_ratio:.6f}"
    )

    # ------------------------------------------------------
    # Most Probable State
    # ------------------------------------------------------

    max_index = np.argmax(probs)

    print(
        f"Most Probable State = "
        f"{states[max_index]}"
    )

    print(
        f"Probability         = "
        f"{probs[max_index]:.6f}"
    )

    # ------------------------------------------------------
    # Optimal States
    # ------------------------------------------------------

    optimal_states = [

        state

        for state in states

        if mis_cost(state, G)
        == optimal_value
    ]

    print(
        "\nOptimal MIS States:"
    )

    print(optimal_states)

    # ------------------------------------------------------
    # Sort Probability Distribution
    # ------------------------------------------------------

    pairs = sorted(
        zip(states, probs),
        key=lambda x: x[1],
        reverse=True
    )

    plot_states = [
        x[0]
        for x in pairs
    ]

    plot_probs = [
        x[1]
        for x in pairs
    ]

    # ------------------------------------------------------
    # Plot Figure
    # ------------------------------------------------------

    fig = plt.figure(
        figsize=(14, 6)
    )

    # ------------------------------------------------------
    # Graph Plot
    # ------------------------------------------------------

    ax1 = fig.add_subplot(1, 2, 1)

    nx.draw(
        G,
        with_labels=True,
        node_size=1500,
        node_color="lightblue",
        font_weight="bold",
        ax=ax1
    )

    ax1.set_title(
        graph_name,
        fontsize=14
    )

    # ------------------------------------------------------
    # Probability Distribution
    # ------------------------------------------------------

    ax2 = fig.add_subplot(1, 2, 2)

    bars = ax2.bar(
        plot_states,
        plot_probs
    )

    # Highlight optimal MIS states

    for i, state in enumerate(plot_states):

        if state in optimal_states:

            bars[i].set_color("red")

    ax2.set_title(
        f"QAOA Probability Distribution (p={P})",
        fontsize=14
    )

    ax2.set_xlabel(
        "Bitstring"
    )

    ax2.set_ylabel(
        "Probability"
    )

    plt.xticks(
        rotation=45
    )

    # Probability labels

    for i, prob in enumerate(plot_probs):

        ax2.text(
            i,
            prob + 0.005,
            f"{prob:.2f}",
            ha="center",
            fontsize=8
        )

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

from scipy.optimize import minimize

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, SparsePauliOp

# ==========================================================
# PARAMETERS
# ==========================================================

LAMBDA = 2
P = 2
N_RESTARTS = 50

# ==========================================================
# GRAPHS
# ==========================================================

graphs = {
    "Path P3": nx.path_graph(3),
    "Cycle C4": nx.cycle_graph(4),
    "Triangle K3": nx.complete_graph(3)
}

# ==========================================================
# BUILD MIS HAMILTONIAN (QUBO → ISING)
# ==========================================================

def build_mis_hamiltonian(G):

    n = len(G.nodes())
    pauli_list = []
    coeffs = []

    # Linear term: -xi = -(1 - Zi)/2
    # contributes constant + Z terms
    for i in range(n):
        z = ['I'] * n
        z[i] = 'Z'
        pauli_list.append("".join(z[::-1]))
        coeffs.append(0.5)  # +Z contribution

    # Edge penalty: λ xi xj = λ/4 (1 - Zi - Zj + ZiZj)
    for u, v in G.edges():

        # ZiZj term
        zz = ['I'] * n
        zz[u] = 'Z'
        zz[v] = 'Z'
        pauli_list.append("".join(zz[::-1]))
        coeffs.append(LAMBDA / 4)

        # Zi term
        zi = ['I'] * n
        zi[u] = 'Z'
        pauli_list.append("".join(zi[::-1]))
        coeffs.append(-LAMBDA / 4)

        # Zj term
        zj = ['I'] * n
        zj[v] = 'Z'
        pauli_list.append("".join(zj[::-1]))
        coeffs.append(-LAMBDA / 4)

    return SparsePauliOp(pauli_list, coeffs=coeffs)


# ==========================================================
# QAOA CIRCUIT (STANDARD COST + MIXER)
# ==========================================================

def qaoa_circuit(G, gammas, betas):

    n = len(G.nodes())
    qc = QuantumCircuit(n)

    qc.h(range(n))

    p = len(gammas)

    for layer in range(p):

        gamma = gammas[layer]
        beta = betas[layer]

        # COST HAMILTONIAN EVOLUTION
        for u, v in G.edges():

            qc.cx(u, v)
            qc.rz(2 * gamma * LAMBDA, v)
            qc.cx(u, v)

        for i in range(n):
            qc.rz(-2 * gamma, i)

        # MIXER
        qc.rx(2 * beta, range(n))

    return qc


# ==========================================================
# EXPECTATION VALUE
# ==========================================================

def expectation(params, H, G, p):

    gammas = params[:p]
    betas = params[p:]

    qc = qaoa_circuit(G, gammas, betas)
    sv = Statevector.from_instruction(qc)

    probs = sv.probabilities()

    n = len(G.nodes())

    value = 0

    for i, prob in enumerate(probs):

        bitstring = format(i, f"0{n}b")[::-1]

        bits = np.array(list(map(int, bitstring)))

        cost = -np.sum(bits) + LAMBDA * sum(
            bits[u] * bits[v] for u, v in G.edges()
        )

        value += prob * cost

    return value


# ==========================================================
# OPTIMIZATION
# ==========================================================

def optimize(G, p):

    H = build_mis_hamiltonian(G)

    best_val = 1e9
    best_params = None

    for _ in range(N_RESTARTS):

        x0 = np.random.uniform(0, np.pi, 2 * p)

        res = minimize(
            expectation,
            x0,
            args=(H, G, p),
            method="COBYLA",
            options={"maxiter": 800}
        )

        if res.fun < best_val:
            best_val = res.fun
            best_params = res.x

    return best_params


# ==========================================================
# MAIN LOOP
# ==========================================================

for name, G in graphs.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    n = len(G.nodes())

    params = optimize(G, P)

    gammas = params[:P]
    betas = params[P:]

    qc = qaoa_circuit(G, gammas, betas)
    sv = Statevector.from_instruction(qc)

    probs = sv.probabilities()

    states = [
        format(i, f"0{n}b")[::-1]
        for i in range(2 ** n)
    ]

    # ======================================================
    # EXPECTATION
    # ======================================================

    exp = 0

    for s, p in zip(states, probs):

        bits = np.array(list(map(int, s)))

        cost = -np.sum(bits) + LAMBDA * sum(
            bits[u] * bits[v] for u, v in G.edges()
        )

        exp += p * cost

    # ======================================================
    # OPTIMAL
    # ======================================================

    optimal = max(
        -np.sum(np.array(list(map(int, s)))) +
        LAMBDA * sum(
            np.array(list(map(int, s)))[u] *
            np.array(list(map(int, s)))[v]
            for u, v in G.edges()
        )
        for s in states
    )

    ratio = exp / optimal

    print(f"Expected MIS Energy = {exp:.6f}")
    print(f"Optimal Energy      = {optimal}")
    print(f"Approx Ratio        = {ratio:.6f}")

    # ======================================================
    # MOST PROBABLE STATE
    # ======================================================

    best = np.argmax(probs)

    print("Most Probable State:", states[best])
    print("Probability:", probs[best])

    # ======================================================
    # PLOT
    # ======================================================

    fig = plt.figure(figsize=(14, 6))

    ax1 = fig.add_subplot(1, 2, 1)
    nx.draw(G, with_labels=True, node_size=1500,
            node_color="lightblue", ax=ax1)
    ax1.set_title(name)

    ax2 = fig.add_subplot(1, 2, 2)

    bars = ax2.bar(states, probs)

    ax2.set_title("MIS-QAOA Probability Distribution")
    ax2.set_xlabel("Bitstring")
    ax2.set_ylabel("Probability")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()
```

```python
from qiskit import QuantumCircuit
from IPython.display import display

# parameters (example)
gamma = 0.7
beta = 0.4
lam = 2

qc = QuantumCircuit(3)

# ==========================================================
# INITIAL STATE (superposition)
# ==========================================================
qc.h([0, 1, 2])

# ==========================================================
# COST HAMILTONIAN (MIS QUBO)
# Graph: 0 -- 1 -- 2 (central node = 1)
# ==========================================================

# edge (0,1)
qc.cx(0, 1)
qc.rz(2 * lam * gamma, 1)
qc.cx(0, 1)

# edge (1,2)
qc.cx(1, 2)
qc.rz(2 * lam * gamma, 2)
qc.cx(1, 2)

# reward terms (-sum x_i)
qc.rz(-2 * gamma, 0)
qc.rz(-2 * gamma, 1)
qc.rz(-2 * gamma, 2)

# ==========================================================
# MIXER LAYER
# ==========================================================
qc.rx(2 * beta, 0)
qc.rx(2 * beta, 1)
qc.rx(2 * beta, 2)

# ==========================================================
# DISPLAY CIRCUIT (BLOCK FORM)
# ==========================================================
display(qc.draw(output="mpl", fold=-1))
```
